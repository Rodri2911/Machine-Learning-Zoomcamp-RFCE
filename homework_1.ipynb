## Q1. Pandas version

```python
import pandas as pd
import numpy as np

pd.__version__
```

## Q2. Records count

```python
# Leemos archivo csv
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"

df_cars = pd.read_csv(url)

print(df_cars.head())
```

```python
total_records = len(df_cars)

print("El numero de registros es:", total_records)
```

## Q3. Fuel types

```python
fuel_types = df_cars['fuel_type'].nunique()
print ("El numero de tipos de gasolina son:", fuel_types)
```

## Q4. Missing values


```python
null_columns = (df_cars.isnull().sum() > 0).sum()
print ("El número de columnas que como minimo tiene un nulo son:", null_columns)
```

## Q5. Max fuel efficiency

```python
cars_asia = df_cars[df_cars['origin'] == 'Asia']
max_fuel_cars_asia = cars_asia['fuel_efficiency_mpg'].max()
print("El combustible máximo en Asia es:", max_fuel_cars_asia)
```

## Q6. Median value of horsepower


```python
median_horsepower = df_cars['horsepower'].median()
mode_horsepower = df_cars['horsepower'].mode()[0]
df_cars['horsepower_new'] = df_cars['horsepower'].fillna(mode_horsepower)
new_median_horsepower = df_cars['horsepower_new'].median()
print("La antigua mediana era: ", median_horsepower," y la nueva mediana es: ", new_median_horsepower)
```

## Q7. Sum of weights

```python
# Carros de Asia
print(cars_asia.head())
new_cars_asia = cars_asia[['vehicle_weight', 'model_year']]
print(new_cars_asia.iloc[:7])
cars_asia_7 = new_cars_asia.iloc[:7]

X = cars_asia_7.values
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)
resultado_final = w.sum()
print("El resultado de la suma de w es:", resultado_final)
```
